```python
import wandb
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Initialize W&B run
run = wandb.init(project="Lab1-visualize-models", name="random-forest-wine")

# 2. Load wine recognition dataset
data = load_wine()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.3, random_state=42
)

# 3. Setup parameters for the model
params = {
    "n_estimators": 100,
    "max_depth": 5,
    "random_state": 42
}
# Log parameters to W&B
wandb.config.update(params)

# 4. Initialize and train the model
model = RandomForestClassifier(**params)
model.fit(X_train, y_train)

# 5. Get predictions and calculate error
pred = model.predict(X_test)
error_rate = 1 - accuracy_score(y_test, pred)
print(f'Test error using Random Forest = {error_rate}')

# 6. Log Summary Metrics
run.summary['Error Rate'] = error_rate

# 7. Generate the Confusion Matrix plot
wandb.sklearn.plot_confusion_matrix(y_test, pred, data.target_names)

# 8. Finish the run
run.finish()
```
